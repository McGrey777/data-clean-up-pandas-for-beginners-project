Load the dataset using pandas, display the first few rows, and explore its structure.

# Find Most Expensive and Cheapest Houses

Identify and print the address and price of the most expensive and cheapest houses in the dataset.

```python
# Find the most expensive house
most_expensive = ds.loc[ds['price'].idxmax()]
print(f"The house located at {most_expensive['address']} is the most expensive, and its price is {most_expensive['price']} USD.")
```

# Find the cheapest house

```python
# Find the cheapest house
cheapest = ds.loc[ds['price'].idxmin()]
print(f"The house located at {cheapest['address']} is the cheapest, and its price is {cheapest['price']} USD.")
```

# Analyze House Sizes

Find and print the largest and smallest houses by area, including their addresses.

```python
# Find the largest and smallest houses by area
largest = ds.loc[ds['area'].idxmax()]
smallest = ds.loc[ds['area'].idxmin()]
print(f"The largest house is located at {largest['address']}, and its area is {largest['area']} square meters.")
print(f"The smallest house is located at {smallest['address']}, and its area is {smallest['area']} square meters.")
```

# Count Unique Populations

Count the number of unique populations in the dataset and print their names.

```python
# Count unique populations
unique_populations = ds['level5'].unique()
print(f"There are {len(unique_populations)} unique populations.")
print(", ".join(unique_populations))
```

# Handle Null Values

Check for null values in the dataset, print the columns with nulls, and remove them if applicable.

```python
# Check for null values
null_columns = ds.columns[ds.isnull().any()]
print(f"Columns with null values: {null_columns}")
ds_cleaned = ds.dropna()
print(f"Dataset size before cleaning: {ds.shape}")
print(f"Dataset size after cleaning: {ds_cleaned.shape}")
```

# Analyze Prices in Specific Population

Calculate and print the average price for houses in 'Arroyomolinos (Madrid)'.

```python
# Calculate average price in 'Arroyomolinos (Madrid)'
arroyomolinos = ds[ds['level5'] == 'Arroyomolinos (Madrid)']
average_price = arroyomolinos['price'].mean()
print(f"The average price in 'Arroyomolinos (Madrid)' is {average_price} USD.")
```

# Visualize Data with Histograms

Plot a histogram of prices for 'Arroyomolinos (Madrid)' and provide a brief analysis of the distribution.

```python
import matplotlib.pyplot as plt

# Plot the histogram of prices
arroyomolinos['price'].hist(bins=20)
plt.title("Histogram of Prices in Arroyomolinos (Madrid)")
plt.xlabel("Price (USD)")
plt.ylabel("Frequency")
plt.show()
```
